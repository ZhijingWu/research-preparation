# 5. Automatic Differentiation

> Source: D2L §2.5

````python
import torch

x = torch.arange(4.0)
x
````

````python
x.requires_grad_(True)  # 等价于x=torch.arange(4.0,requires_grad=True)
x.grad  # 默认值是None
````

````python
y = 2 * torch.dot(x, x)
y
````

````python
y.backward()
x.grad
````

````python
x.grad == 4 * x
````

````python
# 在默认情况下，PyTorch会累积梯度，我们需要清除之前的值
x.grad.zero_()
y = x.sum()
y.backward()
x.grad
````

````python
# 对非标量调用backward需要传入一个gradient参数，该参数指定微分函数关于self的梯度。
# 本例只想求偏导数的和，所以传递一个1的梯度是合适的
x.grad.zero_()
y = x * x
# 等价于y.backward(torch.ones(len(x)))
y.sum().backward()
x.grad
````

````python
x.grad.zero_()
y = x * x
u = y.detach()
z = u * x

z.sum().backward()
x.grad == u
````

````python
x.grad.zero_()
y.sum().backward()
x.grad == 2*x
````

````python
def f(a):
    b = a * 2
    while b.norm() < 1000:
        b = b * 2
    if b.sum() > 0:
        c = b
    else:
        c = 100 * b
    return c
````

````python
a = torch.randn(size=(), requires_grad=True)
d = f(a)
d.backward()
````

````python
a.grad == d / a
````

````python
x = torch.tensor(2., requires_grad=True)

y = x ** 2
u = y.detach()

print(y.requires_grad)
print(y.grad_fn)

print(u.requires_grad)
print(u.grad_fn)
````

````python
x = torch.tensor(3., requires_grad=True)
y = x ** 2
z = y * 5
u = y.detach()

print("x:", x, x.requires_grad, x.grad_fn)
print("y:", y, y.requires_grad, y.grad_fn)
print("z:", z, z.requires_grad, z.grad_fn)
print("u:", u, u.requires_grad, u.grad_fn)

v = u * 10

print(v)
print(v.requires_grad)
print(v.grad_fn)
````

# Day 1：PyTorch Tensor & Autograd

## 1. `requires_grad`

```python
x = torch.tensor(3., requires_grad=True)
```

- `requires_grad=True`：告诉 PyTorch **追踪与该 Tensor 有关的梯度计算**。
- 后续由 `x` 参与的可微运算会自动构建计算图。

---

## 2. `grad_fn`

`grad_fn` 记录这个 Tensor **由什么运算产生**。

```python
x = torch.tensor(3., requires_grad=True)
y = x ** 2

x.grad_fn  # None
y.grad_fn  # PowBackward
```

- `x.grad_fn = None`：`x` 是叶子 Tensor，本身不是通过其他运算产生的。
- `y.grad_fn = PowBackward`：`y` 是通过 `x ** 2` 得到的。

---

## 3. `backward()` 与 `.grad`

```python
loss.backward()
```

从 `loss` 出发，沿计算图进行**反向传播**，计算 loss 对相关变量的梯度。

```python
x.grad
```

表示：

`∂loss / ∂x`

深度学习中通常把最终目标变成一个**标量 loss**，然后调用：

```python
loss.backward()
```

---

## 4. 梯度清零

PyTorch 的梯度默认会**累加**，所以训练时通常需要：

```python
optimizer.zero_grad()
loss.backward()
optimizer.step()
```

流程：

```text
清梯度 → 计算 loss → backward → 得到梯度 → 更新参数
```

---

## 5. `detach()` 与 `torch.no_grad()`

### `detach()`

```python
u = y.detach()
```

切断**某个 Tensor** 与之前计算图之间的梯度关系。

### `torch.no_grad()`

```python
with torch.no_grad():
    y = x ** 2
```

让整个代码块中的运算**不记录梯度**。

记忆：

```text
detach()  → 切 Tensor
no_grad() → 关掉一段代码的梯度记录
```

---

## 6. Autograd 完整流程

```text
requires_grad=True
        ↓
Forward
        ↓
构建计算图
        ↓
得到 scalar loss
        ↓
loss.backward()
        ↓
计算梯度
        ↓
parameter.grad
        ↓
optimizer.step()
        ↓
更新参数
```

## 一句话总结

> **Forward 计算预测和 loss，Backward 计算 loss 对参数的梯度，Optimizer 根据梯度更新参数。**
