## 14. Paper decomposition — Version 1

### Task

> Realize zero-shot inference on text-attributed graphs.

### Bottleneck

> Existing methods either fail to fully utilize graph structure or rely on unreliable LLM outputs when performing zero-shot inference.

### Hypothesis

> Graph-guided bundling can provide richer structural and semantic context, allowing LLMs to generate more reliable supervision signals for training GNNs.

### Intervention

> DENSE dynamically constructs bundles of related nodes based on topology or semantic proximity, queries LLMs to obtain bundle-level labels, uses these labels as supervision to train GNNs, and applies bundle refinement to remove noisy nodes.

### Experiment

> Compare DENSE with existing methods on 10 text-attributed graph datasets, evaluate different LLM backbones, and conduct ablation studies to analyze the contribution of each component.

### Evidence

> DENSE consistently outperforms previous methods across different datasets. Ablation studies show that graph-guided bundling, bundle supervision, and refinement all contribute to the final performance.

DENSE first samples related nodes into bundles based on topology or semantic proximity. Then, it queries the LLM with these bundles to obtain bundle-level supervision.
