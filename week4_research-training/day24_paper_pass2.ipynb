## 4. Loss functions

Record:

Loss 1: 
> $L_{BE}$ = (bundle-level entropy-based / cross-entropy loss)

Purpose:
> It compares the aggregated prediction of all nodes in a bundle with the LLM-generated bundle label. 
This allows the GNN to learn from bundle-level supervision while being more tolerant to noisy or outlier nodes.


Loss 2:

> $L_R$ = (ranking-based loss)

Purpose:

> It penalizes the model when the LLM-generated bundle label is not the highest-probability class in the bundle prediction.


Why are multiple objectives useful?

> $L_{BE}$ encourages the overall bundle prediction to match the LLM label, while $L_R$ additionally ensures that the bundle label is ranked as the most probable class. The two objectives therefore provide complementary supervision.

## 6. Paper decomposition

Problem:

> Perform zero-shot node classification on text-attributed graphs without relying on task-specific labeled data.

Gap:

> Existing methods either fail to fully preserve and utilize graph structural information, or rely on unreliable LLM outputs when making predictions or generating supervision.

Hypothesis:

> Graph-guided bundles can provide richer structural and semantic context, allowing the LLM to generate more reliable bundle-level supervision for training GNNs.

Method:

> First, DENSE uses bundle sampling to construct groups of topologically or semantically related nodes. Then, the text attributes of nodes in each bundle are sent to the LLM, which predicts the main category of the bundle as a bundle-level label. This label is used as supervision to train a GNN through bundle-level objectives. During optimization, bundle refinement is further applied to remove noisy or inconsistent nodes and improve supervision quality.

Experiment:

> First, the paper compares DENSE with multiple baseline families across 10 datasets and evaluates node classification accuracy. Second, it tests DENSE with different LLM backbones to examine whether the method remains effective across different language models. Third, it conducts six ablation variants to study the contribution of bundle sampling, bundle query, the two loss functions, bundle-level supervision, and bundle refinement.

Evidence:

> DENSE consistently achieves stronger prediction accuracy than the compared baselines across the 10 datasets. Results with different LLM backbones show that the framework is not limited to a single LLM. The ablation studies show that removing or replacing major components generally reduces performance, providing evidence that these components contribute to the effectiveness of the full method.


## 7. Researcher's reflection

What is the strongest contribution of DENSE?

> The strongest contribution of DENSE is using graph-guided bundle-level supervision to transfer LLM knowledge to a GNN without treating noisy LLM predictions as reliable node-level labels.


Which component deserves an ablation study?

> Bundle construction deserves further ablation because the quality of bundle supervision directly depends on which nodes are grouped together.
