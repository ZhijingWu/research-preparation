## 1. What is a Graph

图可以表示为：

$$
G=(V,E)
$$

其中：

- $V$: 节点集合 (Nodes)
- $E$: 边集合 (Edges)


与图片不同：

- Image: 固定二维网格
- Graph: 节点数量和连接关系不固定


## 2. Create a Simple Graph

使用 NetworkX 创建一个简单图。

这个例子中：

- 4 个节点
- 4 条边
- 节点之间形成一个环


```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.Graph()

G.add_nodes_from([0, 1, 2, 3])

G.add_edges_from([
    (0, 1),
    (1, 2),
    (2, 3),
    (3, 0)
])

nx.draw(G, with_labels=True)

plt.show()
```

## 3. Node and Edge

### Node

节点表示实体。

例如：

- 社交网络中的用户
- 引用网络中的论文
- 分子中的原子


### Edge

边表示关系。

例如：

- 用户之间的好友关系
- 论文之间的引用关系
- 原子之间的化学键


## 4. Adjacency Matrix

邻接矩阵表示节点之间是否存在连接。

如果：

$$
A_{ij}=1
$$

表示节点 i 和 j 有边。

否则：

$$
A_{ij}=0
$$


```python
import numpy as np

adj = nx.to_numpy_array(G)

adj
```

观察：

邻接矩阵只描述图结构，不包含节点自身的信息。

因此实际 GNN 中通常还需要：

- Node feature
- Edge feature


```python
# 查看节点degree

degree = dict(G.degree())

degree
```

## 5. Degree

Degree 表示一个节点连接多少条边。

例如：

节点0：

连接节点1和节点3

所以：

degree(0)=2


Degree 是最基础的 graph feature。


```python
# 查看节点0的邻居

list(G.neighbors(0))
```

## 7. Why MLP is not enough?


MLP:

输入：

一个独立向量。


但是 Graph:

一个节点不仅有自身feature，还有：

- 邻居
- 连接关系
- 图结构


因此需要：

Graph Neural Network


核心思想：

> 利用邻居信息更新节点表示。
